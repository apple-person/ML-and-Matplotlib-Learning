```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import random
print(matplotlib.__version__)
```

The plot() function draws markers in a diagram.
- Draws a line from point to point.
- takes parameters for specifying points in the diagram.

Parameters
- 1 -- contains points on the x-axis.
- 2 -- contains points on the y-axis.

Points & Markers
- Putting 'o' as the third parameter will create points instead.
- More numbers in list in np.array() creates more points
- To specify a marker, np.array(ypoints, marker = 'o')
- Different markers: 'o', '*', 'x'... full list https://www.w3schools.com/python/matplotlib_markers.asp

```python
xpoints = np.array([0, 2, 4, 6])
ypoints = np.array([3, 9, 6, 12])

plt.plot(xpoints, ypoints, 'o')
plt.show()
```

Without xpoints

```python
ypoints = np.array([3, 8, 1, 10, 5, 7])

plt.plot(ypoints, marker = 'o')
plt.show()
```

You can format strings to make custom markers. The format is as follows: marker|line|color <br>

Different colors are available to use. <br>

There are different ways lines can be drawn:
- '-' : Solid
- ':' : Dotted
- '--' : Dashed
- '-.' : Dashed & dotted
- The parameter lw (linewidth) changes the

Finally, there is additional customization available for markers.
- The parameter ***ms*** changes the marker's size.
- The parameter ***mec*** changes the color of the marker's edge.
- The parameter ***mfc*** changes the color of the marker's face.
- Hexadecimal values are compatible with ***mec*** and ***mfc***

Below is an example with formatting in action.

```python
ypoints = np.array([3, 8, 1, 10, 5, 7])

plt.plot(ypoints, '*-.m', ms = 20, mec = 'r', mfc = 'b', lw = 5)
plt.show()
```

To make labels and titles for the chart:
- plt.xlabel('label') displays *label* under the x-axis
- plt.ylabel('label') displays *label* by the y-axis
- plt.title('label') displays *label* at the top of the graph
<br><br>
You can make multiple lines by adding plt.plot() functions:

```python
y1 = np.array([3, 8, 10])
y2 = np.array([6, 2, 8])

plt.plot(y1)
plt.plot(y2)
plt.xlabel("Time")
plt.ylabel("Production")
plt.title("Production over Time")
plt.show()
```

The axis parameter in the grid() function shows what grids to display. <br>
Legal values: 'x', 'y', 'both' <br>
To define the grid, three parameters need to be specified:
- c (color)
- ls (linestyle)
- lw (line width)

```python
y1 = np.array([3, 8, 10, 6, 7, 5])
plt.plot(y1, marker = 'o')

plt.grid(axis = 'both', c = 'green', ls = '--', lw = 0.5)
```

Displaying multiple plots via the *subplot()* function <br>
*subplot()* takes 3 arguments: row, column, and index. <br><br>
plt.subplot(1, 2, 1)<br>
the figure has 1 row, 2 columns, and this plot is the first plot.<br>
plt.subplot(1, 2, 2)<br>
the figure has 1 row, 2 columns, and this plot is the second plot.<br>

```python
l = []
for i in range(-10, 11):
    l.append(i*i)
y1 = np.array(l)
y2 = np.array([6, 2, 8])

plt.subplot(1, 2, 1)
plt.plot(y1, '*')
plt.subplot(1, 2, 2)
plt.plot(y2)
plt.show()
```

**SCATTER PLOTS**

You can make scatter plots using the *scatter()* function. <br><br>
**Colors**<br>
- Colors are customizable.
- The *color* or *c* argument can be used to set the color.
- To color specific dots, pass the *c* argument as a numPy array.
-  *cmap* arguemnt determines the colormap.
- colormap -- premade set of colors.
- To use *c* with a colormap, set *c* to a numPy array with numbers from 1-100.
<br><br>

**Size**<br>
- The *s* argument controls the size of the dots.
- Should be passed as a numPy array.
<br><br>

**Alpha (Transparency)**<br>
- The *alpha* argument adjusts the transparency of all dots
- Range: 0 (transparent) to 1 (opaque).
<br><br>

```python
plt.xlabel("Age of Car")
plt.ylabel("Speed of Car")

#Set of Data -- 13 cars
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
plt.scatter(x, y, c="#88c999")

#Set of Data -- 15 cars
x = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])
plt.scatter(x, y, c = "red")

plt.show()
```

```python
plt.title("Bubbles!")

#Set of Data -- 13 cars
x = np.random.randint(100, size = (100))
y = np.random.randint(100, size = (100))
colors = np.random.randint(100, size = (100))
sizes = 10 * np.random.randint(100, size = (100))

plt.scatter(x, y, c=colors, s = sizes, alpha = 0.5, cmap = 'plasma_r')
```

**Bar Graphs**
- x-value: Category Names
- y-value: Category Amounts/Values
- Can be displayed horizontally using barh()
- Color can be changed using *c* argument, similarly to scatter plots
- bar() takes *width* argument to change size, barh() takes *height*.

```python
x = np.array(["APPLES", "BANANAS"])
y = np.array([3, 8])
colors = np.array(["red", "yellow"])

plt.bar(x,y, color = colors, width = .5)
plt.show()
```

```python
x = np.array(["APPLES", "BANANAS"])
y = np.array([3, 8])

plt.barh(x,y, height = .5)
plt.show()
```
